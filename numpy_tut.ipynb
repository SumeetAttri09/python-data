```python
arr1 = np.zeros((3,4), dtype = "int64") #prefill
print(arr1, arr1.shape)

arr2 = np.full((3,4), 100) #prefill with val
print(arr2)

arr3 = np.eye(3) #identity matrix
print(arr3, arr3.shape)

arr4 = np.arange(1,11,1) #range(start,end,step)
print(arr4, arr4.shape)
```

```python
#properties
arr = np.array([1,23,4,5])
print(arr)
print(arr.shape) #dimension m*n
print(arr.size) #size of an array
print(arr.dtype) #datatype of an array
print(arr.ndim) #dimention of an array

float_arr = arr.astype(np.float64)
print(float_arr, float_arr.dtype)
```
