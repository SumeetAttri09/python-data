```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data    
```

```python
data = load_data("data_set.json")
print(data)
```

```python
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        #clean the rating
        raw_rating = str(user["Rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["Rating"] = raw_rating

        #Handle missing vals
        raw_age = user.get("Age")
        if(raw_age == None):
            user["Age"] = None

        #Deduplication
        name = user["Name"].strip().lower()
        user["Name"] = name
        if name in unique_users:
            continue

        unique_users.add(name)
        cleaned_data.append(user)

    return cleaned_data  
```

```python
data = clean_data(data)
```

```python
#get meaningfull insights
def get_insights(data):

    #avg Rating
     total_rating = 0
     for user in data:
         total_rating += float(user["Rating"])

     print(f"avg rating = {total_rating / len(data)}")

    #per of poor rating
     poor_rating = 0
     for user in data:
         if(float(user["Rating"])<3):
            poor_rating += 1

     print(f"per of poor rating = {poor_rating / len(data) * 100}%")        
```

```python
get_insights(data)
```
