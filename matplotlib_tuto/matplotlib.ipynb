```python
#non-oscar winner movies
non_oscar_movies = ["qw", "ty", "ko", "op"]
non_oscar_revenue = [5, 15, 24, 19] #in $

plt.plot(years, oscar_revenue,"o--r", label="oscar movies")
plt.plot(years, non_oscar_revenue,"<-y", label="non-oscar movies")

plt.title("oscar movies vs non-oscar movies revenue")

plt.xlabel("years")
plt.ylabel("revenue($)")

plt.legend()

#styling
#plt.style.available
#plt.style.use("dark_background")
plt.grid()
plt.show()
```

```python
#multiple datasets on bar chart
import numpy as np;
oscar_movies = [
    "as",
    "sg",
    "hj",
    "ju",
    "vb"
]
oscar_revenue = [1005, 170, 427, 133, 232] #in $

non_oscar_movies = ["qw", "ty", "ko", "op", "lk"]
non_oscar_revenue = [378, 2788, 829, 185, 275] #in $

years = [2008, 2009, 2010, 2011, 2012]

x = np.arange(len(years))

width = 0.4
plt.bar(x-width/2, oscar_revenue, width, label="oscar-movies") #shifting to left
plt.bar(x+width/2, non_oscar_revenue, width, label="non-oscar movies") #shifting to right

plt.title("Oscar vs non-oscar movies revenue")
plt.xlabel("Year")
plt.ylabel("revenue($)")
plt.legend()

#for horizontal bar graph
#plt.barh(oscar_movies, oscar_revenue)
#plt.xlabel("revenue")
#plt.ylabel("movie name")

plt.xticks(x, years)
```

```python
#scatter plot
people = ["pA", "pB", "pC", "pD", "PE",
          "pF", "pG", "pH", "pI", "pj"]
age = [22,25,30,35,40,45,50,55,60,65]
bp = [110,115,120,122,125,130,135,123,145,150]
# colors = ["green" if x <135 else "red" for x in bp]

plt.scatter(age, bp, cmap="plasma", c=bp)
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("blood pressure")

plt.grid()
plt.colorbar(label="bp")

#annotation
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i]+1, bp[i]+1))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(bp), max(bp)+5)
```
